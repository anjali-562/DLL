```python
import tensorflow as tf
import numpy as np

# Create a tensor
x = tf.constant([[1., 2., 3.], [4., 5., 6.]])

print(x)
print("Shape:", x.shape)
print("Data Type:", x.dtype)

# Tensor operations
print("Addition:\n", x + x)
print("Subtraction:\n", x - x)
print("Multiplication:\n", 5 * x)

# Matrix multiplication
print("Matrix Multiplication:\n", x @ tf.transpose(x))

# Concatenate tensors
result = tf.concat([x, x, x], axis=0)
print("Concatenated Tensor:\n", result)

# Softmax operation
result = tf.nn.softmax(x, axis=-1)
print("Softmax:\n", result)

# Create variables
W = tf.Variable(tf.random.normal([1]))
b = tf.Variable(tf.zeros([1]))

print("Weight:", W)
print("Bias:", b)
```

```python
import torch
import numpy as np

# Create a tensor from a list
data = [[1, 2], [3, 4]]
x_data = torch.tensor(data)

print("Tensor:", x_data)
print("Type:", type(x_data))

# Create tensor from NumPy array
np_array = np.array(data)
x_np = torch.from_numpy(np_array)

print("NumPy Tensor:\n", x_np)

# Create tensors of ones and random values
x_ones = torch.ones_like(x_data)
print("Ones Tensor:\n", x_ones)

x_rand = torch.rand_like(x_data, dtype=torch.float)
print("Random Tensor:\n", x_rand)

# Create a 4x4 tensor of ones and zeros
tensor = torch.ones(4, 4)
tensor1 = torch.zeros(4, 4)

print("Ones:\n", tensor)
print("Zeros:\n", tensor1)

# Concatenate tensors
tensor2 = torch.cat([tensor, tensor1])
print("Concatenated Tensor:\n", tensor2)

# Tensor operations
print("Multiplication:\n", tensor * tensor1)
print("Transpose:\n", tensor.T)

# Add 3 to every element
tensor.add_(3)
print("After Addition:\n", tensor)

# Convert tensor into NumPy array
t = torch.ones(5)
n = t.numpy()

print("Tensor:", t)
print("NumPy Array:", n)
print("Type:", type(n))

# Check GPU availability
if torch.cuda.is_available():
    tensor = tensor.to('cuda')
    print("Device:", tensor.device)
```
